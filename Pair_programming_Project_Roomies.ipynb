> Block 1: Roommates’ names


```python
class Roommate:
    def __init__(self, name):
        self.name = name      # the roommate's name
        self.balance = 0.0    # + means they're owed money, - means they owe

    def __str__(self):
        return self.name + " (balance: $" + format(self.balance, ".2f") + ")"


class Household:
    def __init__(self):
        self.roommates = []   # list of Roommate objects

    def get_names(self):
        names = []
        for r in self.roommates:
            names.append(r.name)
        return names

    def add_roommate(self, name):
        name = name.strip()
        if name == "":
            print("Name can't be blank.")
            return False
        if name in self.get_names():
            print(f"{name} is already in the household.")
            return False
        self.roommates.append(Roommate(name))
        return True

    def find_roommate(self, name):
        for r in self.roommates:
            if r.name == name:
                return r
        return None

    def split_purchase(self, item, cost, payer_name):
        payer = self.find_roommate(payer_name)
        share = cost / len(self.roommates)

        payer.balance += cost          # payer gets credit for the full bill
        for r in self.roommates:
            r.balance -= share         # everyone (payer too) owes one share

        print(f"\n{item}: ${cost:.2f} split {len(self.roommates)} ways = ${share:.2f} each")

    def show_balances(self):
        print("\nCurrent balances:")
        for r in self.roommates:
            print(r)
```

Block 2: Set up the household

```python
home = Household()

while True:
    try:
        num_roommates = int(input("How many roommates live in the household? "))
        if num_roommates > 0:
            break
        print("Enter a number greater than 0.")
    except ValueError:
        print("Please type a whole number, like 3.")

while len(home.roommates) < num_roommates:
    name = input(f"Enter the name of roommate {len(home.roommates) + 1}: ")
    home.add_roommate(name)

print("Roommates:", home.get_names())
```

Block 3: One purchase + balances

```python
item = input("What was purchased? ")

while True:
    try:
        cost = float(input("How much did it cost? $"))
        if cost > 0:
            break
        print("Cost must be more than $0.")
    except ValueError:
        print("Please type a number, like 30 or 12.50.")

payer = input("Who paid? ")
while payer not in home.get_names():
    print("That name isn't in the household. Try again.")
    payer = input("Who paid? ")

home.split_purchase(item, cost, payer)
home.show_balances()
```
